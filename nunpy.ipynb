```python
import numpy as np

arr = np.array([1, 3, 5, 7])
print('numpy.array:', arr)

arr.shape
```

```python
arr2 = np.array([[1,3,5,7], [4,6,8,9]])
print('numpy_array:', arr2)
arr2.shape
```

```python
#dtype => data type ni aniqlash uchun

a = np.array([1, 2, 3])
print(a.dtype)
```

```python
# reshape  => bu arrayning shape'ini o‘zgartiradi,
# elementlar soni esa o‘zgarmaydi.(1D => 2D ...)

a = np.array([1, 2, 3, 4, 5, 6])
b = a.reshape(2, 3)
print('shaped_array:', b)
```

```python
# astype() => bu typenozgartirib beradi
a = np.array([1, 2, 3])
b = a.astype(float)
print('float:', b)
```

```python
# np.zeros() -> berilgan shape'da 0 lardan iborat array yaratadi
# np.ones()  -> berilgan shape'da 1 lardan iborat array yaratadi
# Asosan: dummy data, test qilish va boshlang'ich qiymat berish uchun ishlatiladi.


z = np.zeros((2, 3))
o = np.ones((2, 3))

print("zeros:", z )
print("ones", o )
```

```python
# ndim => array dimension ni aniqlab beradi

np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
]).ndim
```

```python
# arrange bu berilgan qiymatni arange qiberadi (1, 7) => 1 ... 6

a = np.arange(1, 40)
print("a:", a)
```

```python
# sum() => array yegindisini beradi;
# mean()=> ortachasni oberadi
# min() => minimum
# max() => maximum

a = np.array([4, 5, 7, 9, 15])
print("sum:", a.sum())
print("mean:", a.mean())
print("min:", a.min())
print("max:", a.max())
```

```python
# flatten => bu 2D yoki 3D arraylarni => 1D ga ozgartiradi

a = np.array([[1,2], [1,3]])
print(a.flatten())
```

```python
# T (Transpose) => bu array element joyini almashtirib beradi

a = np.array([[1,5], [3,6]])
print("****", a.T)
```

```python
# dot(), @ => bu ikkita arayni bir biriga kopaytiradi

a  = np.array([[3,3], [3,5]])
b  = np.array([[3,3], [3,6]])
print('dot, multiplication', a.dot(b))
```

```python
# random.rand() bu berilgan qiymatdan random array osil qilib beradi only float values 0.0..... ;

a = np.random.rand(2, 4)
print("randoms:", a)
```
