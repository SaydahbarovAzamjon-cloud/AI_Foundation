```python
data_with_missing ={
    "cars": ["BMW", "Volvo", None, "Bugatti", "Ferarri", None, "Porsche"],
    "counts": [None, 7, 5, 4, 3, None, 9]
}

df_missing = pd.DataFrame(data_with_missing)
print(df_missing)
```

```python
df_missing.isnull()
```

```python
df_missing.isnull().sum()
```
